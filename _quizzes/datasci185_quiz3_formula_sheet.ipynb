## 0. Top traps — read before the test
| Trap | Correct pattern / reason |
|---|---|
| `for x in d` gives values? | It gives **keys**; value is `d[x]`. |
| Nested lookup backwards | `people[name]["year"]`: outer key first, then field. |
| Missing / mis-capitalized key | `KeyError`; `"Todd"` and `"todd"` differ. |
| Running best updates only name | Update **both** name and best value inside the `if`. |
| Sum never resets | Put `total = 0` inside the outer loop for each new group/day. |
| `return` inside accumulation loop | Returns on first pass; put it **after** loop to finish all items. |
| `print` instead of `return` | Printed text is not a returned value; no explicit return means `None`. |
| `.apply(check_grade())` | Use `.apply(check_grade)`; pass the function itself. |
| `int("2.5")` | `ValueError`; do not use `int(float(...))` to validate whole-number strings—it truncates. |
| `input()` returns a number? | Always a string; convert with `int` / `float`. |
| Keyword-only argument | `def f(a, b, *, option=False):`; call `f(a, b, option=True)`. |
| Defaults mixed with required inputs | Ordinary required parameters precede ordinary default parameters. |
| `random.shuffle` / `list.append` assignment | Mutates object and returns `None`; call without assignment. |
| `lst = lst + [x]` inside function | Rebinds local name; `.append(x)` mutates caller's list. |
| Integer NumPy array gets `0.1` | May silently truncate to 0; use `dtype=float`. |
| `^` for power | Use `**`; `^` is bitwise XOR. |
| No `break` after first acceptable choice | Loop chooses last acceptable candidate instead. |
| Two decimals required | `f"{value:.2f}"` guarantees two displayed decimal places. |
| Starting max at 0 | Valid only if data assumptions justify it; `float("-inf")` handles negative scores. |

On an exception, remaining statements in that `try` are skipped; the matching handler runs, then execution continues after the handlers. Put `try` **inside** the item loop to keep processing later items. Catch specific errors and report skipped data. An `except` does not undo changes made before the error.

`int("3")` → 3; `float("3.5")` → 3.5; `str(3)` → `"3"`. Parsing an integer does not enforce positivity: extra range rules require separate validation. The practice solution assumes the supplied quantity choices.

````python
def ask_positive_integer():
    while True:
        text = input("Enter a positive whole number: ").strip()
        try:
            number = int(text)
        except ValueError:
            print("Please enter a whole number.")
            continue
        if number <= 0:
            print("Please enter a number above zero.")
            continue
        return number

# Call ask_positive_integer() only when you want an interactive prompt.
````

The set-up cell below defines `ga_temps`, a dictionary of five Georgia cities. Each value is itself a dictionary that maps each day of the week to that city's high temperature (°F) on that day. Using `for` loops, compute the average high temperature on each day of the week across the cities in `cities_to_be_averaged`. Print one line per day in the form "Average high on {day}: {temp} °F", rounded to 2 decimal places.


For exampl,  if the selected cities were Atlanta, Savannah, and Athens, the average high on Monday would be $(78 + 84 + 76)/3 \approx 79.33$ and your code should print "Average high on Monday: 79.33 °F"

````python
###### Set-up code (do not change)
import random

ga_temps = {
    "Atlanta":  {"Monday": 78, "Tuesday": 81, "Wednesday": 75, "Thursday": 72, "Friday": 74, "Saturday": 79, "Sunday": 82},
    "Savannah": {"Monday": 84, "Tuesday": 86, "Wednesday": 83, "Thursday": 80, "Friday": 81, "Saturday": 85, "Sunday": 87},
    "Augusta":  {"Monday": 80, "Tuesday": 83, "Wednesday": 78, "Thursday": 74, "Friday": 76, "Saturday": 81, "Sunday": 84},
    "Macon":    {"Monday": 82, "Tuesday": 84, "Wednesday": 79, "Thursday": 76, "Friday": 77, "Saturday": 83, "Sunday": 85},
    "Athens":   {"Monday": 76, "Tuesday": 79, "Wednesday": 73, "Thursday": 70, "Friday": 72, "Saturday": 77, "Sunday": 80},
}

days_of_week = ["Monday", "Tuesday", "Wednesday", "Thursday", "Friday", "Saturday", "Sunday"]

num_cities = random.randint(2, 4)
selected = random.sample(["Atlanta", "Savannah", "Augusta", "Macon", "Athens"], k = num_cities)

cities_to_be_averaged = {}
for city in selected:
    cities_to_be_averaged[city] = ga_temps[city]

# cities_to_be_averaged # uncomment if you want to see the cities selected
###### (end of set-up code)
````

The code below computes how many years it takes for an initial population of 1,000 to double in size, where it grows by `growth_rate` each year.

Turn this code into a function `years_to_double(growth_rate)` that returns the number of years required for an initial population of 1000 to double in size. Then use your function to print the doubling time for growth rates of 1%, 2%, 5%, and 10%. You should get 70, 36, 15, and 8 years.

````python
population = 1000
growth_rate = 0.03
years = 0

while population < 2000:
    population = population * (1 + growth_rate)
    years = years + 1

print(years)
````

````python
# SOLUTION
def years_to_double(growth_rate):
    population = 1000
    years = 0
    while population < 2000:
        population = population * (1 + growth_rate)
        years = years + 1
    return years

for rate in [0.01, 0.02, 0.05, 0.10]:
    print(f"Growth rate {rate}: {years_to_double(rate)} years")
````

An online fruit stand stores its prices in the dictionary `prices`. Customers type their orders into a web form. Each order is stored as a dictionary whose keys are items and whose values are quantities. Because customers type the quantities, they are stored as strings, and some of them aren't whole numbers.

Write a function `order_total(order, prices)` that returns the total cost of an order. Skip any item that can't be added to the total, and print a message explaining why:

- If the store doesn't sell the item, print a message like "Sorry, we don't sell {item}.
- If the quantity isn't a whole number, print a message like "Sorry, {string} isn't a valid quantity".

Your function should detect both problems using `try/except` statements, not `if` statements. If an item has both problems, printing either message is fine.

For example, 
```python
order = {"apple": "4", "kiwi": "2", "banana": "lots", "cherry": "1"}
total = order_total(order, prices)
```
should print the strings "Sorry, we don't sell kiwi", followed by "Sorry, lots isn't a valid quantity", and assign the value 5.0 to "total".

The set-up cell below generates a random order. Run it several times to test your function on different orders.

````python
# set up cell DO NOT CHANGE
import random

prices = {"apple": 0.50, "banana": 1.25, "cherry": 3.00, "mango": 1.35, "pear": 0.75}

all_items = ["apple", "banana", "cherry", "mango", "pear", "kiwi", "coconut", "lime"]
all_quantities = ["1", "2", "3", "4", "5", "6", "a dozen", "lots", "3 or 4"]

num_items = random.randint(4, 6)
items_ordered = random.sample(all_items, k = num_items)

order = {}
for item in items_ordered:
    order[item] = random.choice(all_quantities)

order
````

````python
# SOLUTION

def order_total(order, prices):
    total = 0
    for item in order.keys():
        try:
            total = total + prices[item] * int(order[item])
        except KeyError:
            print(f"Sorry, we don't sell {item}")
        except ValueError:
            print(f"Sorry,{order[item]} isn't a quantity")
    return total


print("Total:", order_total(order, prices))
````

**Question (F) - (material from Lab 5)** 



In the cell below, a function is defined that takes two points as input and computes their Euclidean distance. The two points passed into the function are assumed to be in Cartesian coordinates $(x,y)$. We want to create a function that works for either Cartesian coordinates $(x, y)$ or polar coordinates $(r, \theta)$. Recall the relationship between the two coordinate systems:

$$ x = r \cos(\theta),\quad y = r\sin(\theta)\qquad\qquad \mathrm{(*)} $$


Write a function, `cord_dist`, that:

 - Takes as arguments the coordinates of two points, each point given as a tuple, and returns the distance between the two points.
 - In addition, it should take a third **keyword-only** with the default value `polar = False`. When "polar" has its default value of `False`, then it should work the same as the function "euclidean_dist" defined in the cell below.

- If the user passes `polar = True`, this indicates that the points passed to the function are tuple containing polar coordinates $(r, \theta)$. So the tuples must be converted to $(x, y)$ coordinates before computing the distance.

 - Allow the user of the function to input the two tuples positionally. (So only `polar` should be restricted to keyword-only).

**Note:** Feel free to call the function `euclidean_dist` inside the one you write.

````python
# given 
import numpy as np

def euclidean_dist(point1, point2):
    # extract coordinates from tuples
    x1 = point1[0]
    y1 = point1[1]
    x2 = point2[0]
    y2 = point2[1]
    return np.sqrt((x1-x2)**2+(y1-y2)**2)

# demonstration of function
first_point = (5, 3)
second_point = (-1, 2)

dist = euclidean_dist(first_point, second_point)
print(dist)
````

````python
# SOLUTION

def cord_dist(coord1, coord2, *, polar = False):
    if polar == True:
        coord1 = (coord1[0]*np.cos(coord1[1]), coord1[0]*np.sin(coord1[1]))
        coord2 = (coord2[0]*np.cos(coord2[1]), coord2[0]*np.sin(coord2[1]))

    return euclidean_dist(coord1, coord2)


point_1 = (1.41, 0.79)
point_2 = (6.4, 0.9)
print(cord_dist(point_1, point_2, polar=True))

point_a = (1, 1)
point_b = (4, 5)
print(cord_dist(point_a, point_b))
````

## 7. Silent errors, assertions, and tests — Lecture 13
**Silent error:** valid code runs, but gives the wrong answer. `try/except` cannot catch incorrect logic by itself.

```python
assert 0 <= probability <= 1, "probability outside [0,1]"
assert len(roster) == expected_size, "wrong roster size"
for name in roster:
    assert 0 <= roster[name]["lab_skills"] <= 10, f"bad score for {name}"
```
`assert condition, message`: true → no output; false → stop with `AssertionError`. Put **preconditions** at function entry and **postconditions** before return. Use `try/except` and explicit checks for routine bad user input.

**Debugging recipe:** state expected answer → shrink failing example → inspect intermediate values / step through → change one thing → rerun → save an independent test.

| Bug / situation | Check / fix |
|---|---|
| Birthday loop compares a person to themself | Inner loop must be `range(i + 1, room_size)`. |
| Return False on first nonmatch | Return True when match found; return False **after all pairs**. |
| Running max never updated | Update best score with best name. |
| Selection does not stop | Add `break` after first person above benchmark. |
| Mean looks fine despite wrong stopping | Return chosen position too; test position, not only mean. |
| Probability looks plausible | Also test known answers; range checks alone do not prove correctness. |
| Missing scores encoded `-999`, `999`, or `0` | Check source meaning; range checks miss an in-range missing code. |
| Dropping missing observations | Report kept/dropped counts; dropping can bias averages. |

**Independent tests:** one person cannot share a birthday; 361 people must share a birthday in the lecture's **360-day model** (12 months × 30 days). These do not depend on random luck. For positive lab skills and N=0, the first candidate must be chosen. Random averages require tolerances and can occasionally fail by chance; more trials stabilize estimates, not each individual outcome.

`math.isclose(actual, expected)` for float comparison; `abs(actual - expected) < tolerance` for explicit absolute tolerance. Avoid equality-based stopping on floats; an integer step counter is often clearer.

````python
def clean_roster(roster):
    clean = {}
    for name in roster:
        score = roster[name]["lab_skills"]
        if 0 <= score <= 10:  # assumes documented valid score range
            clean[name] = roster[name]
    print(f"Kept {len(clean)}; dropped {len(roster) - len(clean)}")
    return clean

# A new outer dictionary still shares the inner dictionaries with roster.
````

## 8. Lab partner simulation, plotting, and input — Lab 6
**Strategy:** shuffle names → observe first N and record benchmark → choose first later score strictly above benchmark → otherwise choose last candidate. This example passes the roster explicitly; Lab 6 expects `run_one_trial(N)` using the global `lab_partners` roster.

````python
def trial_with_roster(N, roster):
    assert len(roster) > 0, "roster must not be empty"
    assert isinstance(N, int) and 0 <= N < len(roster), "invalid N"
    order = list(roster.keys())
    random.shuffle(order)
    benchmark = 0  # assumes all valid skills are strictly positive
    for position in range(N):
        skill = roster[order[position]]["lab_skills"]
        if skill > benchmark:
            benchmark = skill
    chosen_name = order[-1]  # fallback if nobody beats benchmark
    for position in range(N, len(order)):
        name = order[position]
        if roster[name]["lab_skills"] > benchmark:
            chosen_name = name
            break
    return chosen_name, roster[chosen_name]["lab_skills"]

demo_roster = {"A": {"lab_skills": 2}, "B": {"lab_skills": 6},
               "C": {"lab_skills": 9}, "D": {"lab_skills": 4}}
mean_skills = []
for N in range(len(demo_roster)):
    chosen_skills = []  # reset results for each strategy
    for trial in range(100):
        name, skill = trial_with_roster(N, demo_roster)
        chosen_skills.append(skill)
    mean_skills.append(np.mean(chosen_skills))
print(mean_skills)  # random results vary
````

```python
# Plot average outcome for each strategy
plt.bar(x=np.arange(len(mean_skills)), height=mean_skills)
plt.xlabel("N")
plt.ylabel("average chosen skill")
plt.title("Comparing strategies")
plt.show()

# Distribution of individual outcomes for ONE N
plt.hist(x=chosen_skills, bins=20)
plt.xlabel("chosen skill")
plt.ylabel("number of trials")
plt.title(f"Chosen skills (std = {np.std(chosen_skills):.2f})")
plt.show()
```
`np.mean(values)` = typical outcome; `np.std(values)` = spread of outcomes (default population SD, `ddof=0`). `np.argmax(mean_skills)` returns the **index** of largest mean; that equals N here because N starts at 0 and increases by 1. More trials make estimated averages steadier; nearby winners may still switch.

**Normalized input:** `answer = input("pick or pass? ").strip().lower()`. `.strip()` removes leading/trailing whitespace; `.lower()` lowercases. Loop until response is in `["pick", "pass"]`; invalid response should not advance to the next candidate. In the game, automatically take the last candidate if all others were passed.
