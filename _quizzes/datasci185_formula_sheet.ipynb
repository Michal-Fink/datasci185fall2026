### Bar charts — and why x-axis type matters

`plt.bar` for vertical bars, `plt.barh` for horizontal. From Lecture 3: if your x-values are **strings** (e.g. `"11", "12", "1", "2"`), they plot in the order given/sorted as text. If they're **numbers**, they plot in numeric order. Convert with `int()`/`float()` first if you need numeric ordering.

```python
months_as_numbers = [11, 12, 1, 2]
sales = [100, 250, 80, 300]

plt.bar(x=months_as_numbers, height=sales)
plt.title("x-values stored as integers")
plt.xlabel("month number")
plt.ylabel("sales")
plt.show()

# horizontal version — first arg is the labels, second is the bar length
plt.barh(["red", "green", "blue"], [10, 25, 15])
plt.show()
```

## 5. Variables — swapping

This is a classic trap: `b = a; a = b` does **not** swap — by the time the second line runs, `b` already equals `a`, so `a = b` just reassigns `a` to itself. Use a temp variable, or Python's tuple-swap trick.

```python
a = "left"
b = "right"

# WRONG (does not swap):
# b = a
# a = b

# CORRECT — temp variable:
temp = a
a = b
b = temp

# CORRECT — Pythonic one-liner (does this in one step, no temp needed):
a, b = b, a
```

### Indexing — positive and negative

Python indexing starts at **0**. Negative indices count from the end: `-1` is the last element, `-2` second-to-last, etc. Indexing works the same way on nested lists — just chain the brackets.

```python
list_colors = ["red", "green", "yellow", "indigo"]
list_colors[3]    # "indigo" — 4th element, index 3
list_colors[-1]   # "indigo" — same thing, counting from the end

# Nested list example
favorite_movies = ["Lucy", "Lego Movie", "The Hunger Games", "Ghost Dog"]
another_list = [7, 5, favorite_movies, "20", 20]

last_movie = another_list[2][-1]   # index into the nested list, then grab its last item
print(last_movie)  # "Ghost Dog"
```

### Copying a list — the #1 gotcha

`new_list = old_list` does **not** make a copy — both names point to the *same* list, so editing one edits both. Use `.copy()` when you need an independent copy.

```python
list_of_floats = [2.3, -1.0, 5.1, 9.9]

not_a_copy = list_of_floats          # same list, two names
list_of_floats[0] = "Nothing"
print(not_a_copy)                    # ['Nothing', -1.0, 5.1, 9.9] -- changed too!

list_of_floats2 = [2.3, -1.0, 5.1, 9.9]
real_copy = list_of_floats2.copy()   # independent copy
list_of_floats2[0] = "Nothing"
print(real_copy)                     # [2.3, -1.0, 5.1, 9.9] -- unaffected
```

## 8. Type conversion (Lecture 3)

`str()`, `int()`, `float()` convert deliberately. **Big trap:** `+` on two strings concatenates, it does NOT add numbers — you must convert first.

```python
a = "10"
b = "2"

print(a + b)              # "102"  <- string concatenation, NOT addition
print(int(a) + int(b))    # 12     <- convert first, then add

# convert the numeric result back to a string if needed
c = str(int(a) + int(b))  # "12"
```

## 9. Comparisons and Booleans (Lecture 3 & 4)

### Booleans vs. the strings "True"/"False"

`True`/`False` (no quotes) are `bool` objects. `"True"`/`"False"` (with quotes) are just strings — don't confuse them.

```python
print(type(True))     # <class 'bool'>
print(type("True"))   # <class 'str'>
```

### Comparison operators

`>`, `<`, `>=`, `<=`, `==` (equal), `!=` (not equal). **Don't confuse `=` (assignment) with `==` (comparison).** Chained comparisons work: `10 < price <= 20`. Strings/lists compare too — alphabetical for strings, element-by-element (order matters) for lists.

```python
price = 24.5
print(price > 10 and price <= 20)   # False
print(10 < price <= 20)             # same thing, chained -- False

print("Nathan" < "Peter")           # True -- alphabetical order
print(["a", "b"] == ["a", "b"])     # True -- same order, same elements
print(["a", "b"] == ["b", "a"])     # False -- order matters
```

### Floating point equality — the trap

`0.1 * 7 == 0.7` is `False` because of floating-point rounding error, even though they look equal. **Never test two floats with `==`.** Use `math.isclose()`, or check that the absolute difference is tiny.

```python
x = 0.1 * 7
y = 0.7

print("x and y are equal:", x == y)          # False — floating point error

# Fix option 1: round both sides before comparing
print("rounded equal:", round(x, 5) == round(y, 5))   # True

# Fix option 2: math.isclose (best practice for floats)
import math
print("isclose:", math.isclose(x, y))        # True

# Fix option 3: check the absolute difference is tiny
print("small diff:", abs(x - y) < 1e-13)     # True
```

### Logical operators — `and`, `or`, `not`, `^`

`and`: both must be True. `or`: at least one True (inclusive). `^`: exactly one True (exclusive or). `not`: flips it. Use parentheses to keep chains readable.

```python
p, q = True, False

print(p and q)   # False
print(p or q)    # True
print(p ^ q)     # True -- exactly one is True
print(not p)     # False

age = 21
print((age > 18) and (not False))   # parentheses for clarity -> True
```

### The `in` keyword

Checks membership: a character in a string, an item in a list, a value in a numpy array. Case-sensitive.

```python
# substring in a string
sentence = "The Federal Reserve makes economic forecasts"
print("economic" in sentence)   # True
print("federal" in sentence)    # False -- case-sensitive (capital F in sentence)

# item in a list
majors = ["Biology", "Economics", "Data Science"]
print("Biology" in majors)      # True
```

## 10. Built-in functions (Lecture 4)

`type()`, `abs()`, `len()`, `round()`, `str()`. Inputs = arguments, outputs = return values. `round()` takes an optional 2nd argument (digits) and can be called with keyword args — but once you use a keyword arg, every arg after it must also be a keyword (`round(3.14, ndigits=2)` is fine, `round(number=3.14, 2)` is a `SyntaxError`).

```python
print(type(3.14))       # <class 'float'>
print(abs(-10))         # 10
print(len([2, 3, 4]))   # 3  (len fails on a plain number/int)

print(round(3.1415))       # 3     -- no 2nd arg -> rounds to an int
print(round(3.1415, 3))    # 3.142 -- 2nd arg = digits to keep
print(round(3.1415, ndigits=3))   # same, using the keyword name
```

## 11. `math`, `statistics`, and `numpy`

`math` and `statistics` work on **single numbers/one list at a time**. `numpy` (`np`) is built for doing the same operation across **many values at once** — `math.sqrt(a_list)` errors, `np.sqrt(a_list)` doesn't.

```python
import math, statistics, numpy as np

# math -- single values
print(math.sqrt(25))       # 5.0
print(math.factorial(5))   # 120
print(math.pi)              # 3.141592653589793 (a value, not a function -- no parens)

# statistics -- summarize a list
nums = [-5, -0.1, 1.2, 3, 5.3, 11, 25]
print(statistics.mean(nums))
print(statistics.median(nums))
print(statistics.stdev(nums))

# numpy -- operates on a whole list/array at once
side_areas = [4, 9, 16, 25]
print(np.sqrt(side_areas))   # [2. 3. 4. 5.]  -- math.sqrt(side_areas) would error

print(np.sum([1, 2, 3, 4]))      # 10
print(np.mean([1, 2, 3, 4]))     # 2.5
print(np.cumsum([1, 2, 3, 4]))   # [ 1  3  6 10]  -- running total

quiz_scores = np.array([7.5, 9.0, 4.5, 8.0, 6.5])
print(np.argmax(quiz_scores))    # 1  -- INDEX of the max value, not the value itself
print(np.min(quiz_scores))       # 4.5 -- the value itself
```
